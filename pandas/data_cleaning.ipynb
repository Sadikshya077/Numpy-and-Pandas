# Data Cleaning

```python
'''
1. Handle missing / null values
    a. fill null values
        -> .fillna()
            continuous:
                - mean
                - median
                -> skewness -> .skew()
                 - 0-0.5 (fill mean value)
                 - <0 (negative skewness) or >0.5(positive skewness) (fill median value)
                 or
                -> calculate mean and median value of a column
                 - if both mean and median value are equal or almost equal, fill with mean
                 - else fill with median like for data with many outliers
            discrete / categorical:
                - mode
            continuous / discrete / categorical:
                - custom record
    b. drop null values
        -> .dropna()
        
2. Remove unwanted data
    -> remove whitespace
        - .strip(), .lstrip(), .rstrip()
    -> remove regular expressions (regex) if needed / remove unwanted texts
        - .replace()
        
3. Change Datatypes
    .astype()
    
4. Handle duplicate records
    -> find duplicate records
        -> .duplicated() / .duplicated(subset) for checking specific column
    -> remove duplicate records
        -> .drop_duplicates()
'''
```

```python
import pandas as pd
```

## Drop Null Values

```python
# -------------Row wise drop------------------------------------
# df.dropna() -> does temporary change and doesn't change the df
# df = df.dropna() -> one technique to change the df permanently
# df.dropna(inplace=True) another technique
```

```python
'''
- dropna(axis=0/1, how='any'/'all', inplace= True/ False)

- axis -> 0 -> Row / 1-> Column
- how 
    -any -> data should have any one null value
    -all -> data must have all null values
- inplace
    -True -> permanent change to the copy of excel data i.e df
    -False -> temporary change to the copy of excel data i.e df
    (excel file doesn't change unless we replace it with the df)
- default 
    dropna() -> axis = 0, how = 'any', inplace = False
'''
```

```python
df.info()
```

```python
df.dropna(axis=0,how='any',inplace=False)
df
```

```python
df.head()
```

# Fill Null Values

```python
df=pd.read_csv('Dataset/Employees.csv')
```

```python
df.isnull().sum()
```

### Fill commission values

```python
commission_mean=round(df['COMMISSION_PCT'].mean(),2)
commission_median=df['COMMISSION_PCT'].median()
commission_skew=df['COMMISSION_PCT'].skew()
```

```python
print(commission_mean)
print(commission_median)
print(commission_skew)
```

```python
# df['COMMISSION_PCT'].fillna(commission_mean,inplace=True)
# avoid using this method 
```

```python
df.fillna({'COMMISSION_PCT':commission_mean},inplace=True)
```

```python
df.isnull().sum()
```

```python
'''
df.fillna({'COMMISSION_PCT':0,},inplace=True)
to fill with custom value=0
'''
```

```python
df['COMMISSION_PCT'].unique()
```

### Fill Manager and Department Values

```python
# we should keep zero index since some columns can have multiple modes
# by keeping [0], it takes first mode 
mgr_mode=df['MANAGER_ID'].mode()[0]
dep_mode=df['DEPARTMENT_ID'].mode()[0]
```

```python
print(mgr_mode)
print(dep_mode)
```

```python
df.fillna({'MANAGER_ID':mgr_mode},inplace=True)
df.fillna({'DEPARTMENT_ID':dep_mode},inplace=True)
```

```python
df.to_csv('cleaned_employee.csv',index=False)
df.to_excel('cleaned_employee.xlsx',index=False)
df.to_json('cleaned_employee.json',indent=4)
```
