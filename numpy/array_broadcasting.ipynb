```python
import numpy as np
```

```python
'''
Rules of Broadcasting
----------------------------------
1. The shape of two arrays should be equal to perform any operations
2. If not equal then the shape of one array should have either equal rows and column 1
    or equal column and row 1
'''
```

```python
arr_1=np.array([
    [1,2,3],
    [4,5,6],
    [6,7,8]
])

arr_2=np.array([
    [2,5,6],
    [6,5,4],
    [1,2,3]
])

arr_3=np.array(
    [4,8,9]
)
```

```python
arr_1 + arr_2
```

```python
arr_1 + arr_3
```
