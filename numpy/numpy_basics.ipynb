# Numeric Python

```python
import numpy as np
```

```python
a=[2,4,6]
b=[1,3,5]
a
```

```python
a_array=np.array(a)
print(type(a_array))
print(a_array)
```

```python
b_array=np.array(b)
print(b_array)
```

```python
addition_list=a+b
addition_numpy=a_array+b_array
print(addition_list)
print(addition_numpy)
```

```python
divide_float=a_array/b_array
divide_int = a_array // b_array
print(divide_float)
print(divide_int)
```

```python
# multi-Dim array
# all the list should have same shape
multi_d=np.array([
    [2,4,6],
    [1,3,5],
    [5,6,7]
])
print(multi_d)
print(multi_d.shape)
```
