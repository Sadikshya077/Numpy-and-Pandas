# Slicing

```python
import numpy as np
```

### 1D array

```python
num_1d=np.arange(1,11)
num_1d
```

```python
num_1d[2:4]
```

### 2D array

```python
num=np.arange(1,51)
num
```

```python
num_2d=num.reshape(10,5)
num_2d
```

```python
# without using numpy
for row in num_2d[1:3]:
    print(row[1:3])
```

```python
# we can do directly by: num_2d[row,column]
num_2d[1:3,1:3]
```
