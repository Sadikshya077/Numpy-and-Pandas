# Conditional Filtering

```python
df=pd.read_csv('Dataset/Car_Sales.csv')
df.head()
```

```python
df[df['year']==2017][['name','year','selling_price']]
```

```python
df.loc[(df['selling_price']>=300000) & (df['selling_price']<=600000), ['name','year','selling_price']]
```

```python
df[(df['selling_price']>=300000)&(df['selling_price']<=600000)]
```

```python
df[(df['selling_price']>=300000)&(df['selling_price']<=600000)][['name','year','selling_price']]
```

```python
df[df['selling_price'].between(300000,600000)]
```

```python
df[df['selling_price'].between(300000,600000)][['name','year','selling_price']]
```

```python
# to find unique values
df['fuel'].unique()
```

```python
# to find number of unique values in a column
df['fuel'].nunique()
```

```python
# find cars whose emmision type is petrol or CNG
car_fuel=df[(df['fuel']=='Petrol') | (df['fuel']=='CNG')]
car_fuel
```

```python
car_fuel_in=df[df['fuel'].isin(['Petrol','CNG'])]
```

```python
car_fuel_in
```

```python
''' find car types whose transmission mode is Automatic, fuel type is Petrol/Electric and 
which is driven around 20000 to 35000 range
'''
```

```python
car_types=df[(df['transmission']=='Automatic') & 
             ((df['fuel']=='Petrol') | (df['fuel']=='Diesel')) &
             # (df['km_driven'].between(20000,35000))]
             ((df['km_driven']>=20000) & (df['km_driven']<=35000))]
car_types
```

```python
'''
Data types
-----------
-int , str, bool, None, complex, float, datetime

Types of data
---------------
-Numeric:
   -Discrete
   -Continuous

-Categorical
   -nominal -> category which cannot be compared
   -ordinal -> category which can be compared
'''
```

```python
temp=df[(df['selling_price']>=300000) & (df['selling_price']<=600000)][['name','year','selling_price']]
temp
```

```python
temp = df[df['selling_price'].apply(lambda x: (x>=300000 & x<=600000))]
temp                                      
```
