# Reshape and Ravel

```python
import numpy as np
```

```python
numbers=np.arange(1,61)
numbers
```

```python
'''
reshape -> converts the array into any dimension.
ravel/ flatten in sklearn -> converts the array into one dimensional
'''
```

```python
reshape_numbers=numbers.reshape(10,6)
reshape_numbers
```

```python
ravel_numbers=reshape_numbers.ravel()
ravel_numbers
```
