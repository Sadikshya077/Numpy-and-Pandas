# Data Cleaning

```python
'''
1. Handle missing / null values
    a. fill null values
        -> .fillna()
            continuous:
                - mean
                - median
                -> skewness -> .skew()
                 - 0-0.5 (fill mean value)
                 - <0 (negative skewness) or >0.5(positive skewness) (fill median value)
                 or
                -> calculate mean and median value of a column
                 - if both mean and median value are equal or almost equal, fill with mean
                 - else fill with median like for data with many outliers
            discrete / categorical:
                - mode
            continuous / discrete / categorical:
                - custom record
    b. drop null values
        -> .dropna()
        
2. Remove unwanted data
    -> remove whitespace
        - .strip(), .lstrip(), .rstrip()
    -> remove regular expressions (regex) if needed / remove unwanted texts
        - .replace()
        
3. Change Datatypes
    .astype()
    
4. Handle duplicate records
    -> find duplicate records
        -> .duplicated() / .duplicated(subset) for checking specific column
    -> remove duplicate records
        -> .drop_duplicates()
'''
```

```python
import pandas as pd
```

```python
df=pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
# check null values
df.isnull().sum() # shows column wise null values count
```

```python
df.isnull().sum().sum() # sum every null values from all columns
```

## Drop Null Values

```python
# -------------Row wise drop------------------------------------
# df.dropna() -> does temporary change and doesn't change the df
# df = df.dropna() -> one technique to change the df permanently
# df.dropna(inplace=True) another technique
```

```python
'''
- dropna(axis=0/1, how='any'/'all', inplace= True/ False)

- axis -> 0 -> Row / 1-> Column
- how 
    -any -> data should have any one null value
    -all -> data must have all null values
- inplace
    -True -> permanent change to the copy of excel data i.e df
    -False -> temporary change to the copy of excel data i.e df
    (excel file doesn't change unless we replace it with the df)
- default 
    dropna() -> axis = 0, how = 'any', inplace = False
'''
```

```python
df.info()
```

```python
df.dropna(axis=0,how='any',inplace=False)
df
```

```python
df.head()
```

```python
# Drop any column
# df.drop(columns='EMAIL') for single column
df.drop(columns=['COMMISSION_PCT','PHONE_NUMBER']) # for multiple columns
```

```python
# drop any row
df.drop(index=[0,1,2]) # for multiple rows
```

```python
df.drop(index=4) # for single row
```

## Find duplicate records

```python
duplicate_data= df[df.duplicated()]
duplicate_data
```

```python
duplicate_data= df[df.duplicated(subset=['EMPLOYEE_ID'], keep=False)]
#if we remove keep=False then it will give the duplicated data once not with it's duplicates
duplicate_data
```

## Remove duplicates

```python
df.drop_duplicates(subset='EMPLOYEE_ID', keep='last',inplace=True)
# keeps the last data and remove all duplicates above it
```
