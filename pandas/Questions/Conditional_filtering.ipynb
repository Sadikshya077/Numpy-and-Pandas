## Question 1: Single Boolean Condition

```python
'''
The Head of IT needs to review the roster for his technical team. Write a Pandas statement to filter
and isolate all rows where the Department is exactly "IT/IS".
'''
```

```python
df['Department']
```

## Question 2: Multi-Condition Filtering using AND (&)

```python
'''
HR wants to find potential flight-risk or disengaged employees in production roles. 
Write a query to find all employees who are in the "Production" department AND
have an EmpSatisfaction score less than 3.
'''
```

```python
df.head()
```

```python
filtered=df[(df['Department']=='Production') & (df['EmpSatisfaction']<3)]
filtered
```

## Question 3: Multi-Condition Filtering using OR (|)

```python
'''
The talent development team wants to reward recruitment channels. 
Filter for records where the employee was sourced via "Glassdoor" OR via "Employee Referral".
'''
```

```python
df.columns
```

```python
df['RecruitmentSource'].unique()
```

```python
filtered=df[(df['RecruitmentSource']=='Glassdoor') | (df['RecruitmentSource']=='Employee Referral')]
filtered
```

## Question 4: Numeric Ranges and Column Slicing

```python
'''
Management is reviewing payroll and attendance anomalies. Filter for all employees who have a
Salary greater than 100,000 AND have logged more than 10 Absences. Display only their
Employee_Name, Position, and ManagerName.
'''
```

```python
df['Salary']
```

```python
df['Absences']
```

```python
filtered=df[(df['Salary']>100000) & (df['Absences']>10)][['Employee_Name','Position','ManagerName']]
filtered
```
